```python
# ============================================================
# ECLAT - PHARMA A/B TEST DATASET
# UNSUPERVISED LEARNING
# ============================================================

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from itertools import combinations
from collections import defaultdict


# ============================================================
# 1. LOAD DATASET
# ============================================================

df = pd.read_csv("pharma_ab_test_data.csv")

print("Dataset Shape:", df.shape)

print("\nFirst 5 Rows:")
print(df.head())


# ============================================================
# 2. CREATE TRANSACTION ITEMS
# ============================================================

transactions = []

for _, row in df.iterrows():

    transaction = []

    transaction.append("Group_" + str(row["group"]))
    transaction.append("Pack_" + str(row["pack_color"]))
    transaction.append("Age_" + str(row["age_group"]))
    transaction.append("Gender_" + str(row["gender"]))
    transaction.append("Device_" + str(row["device_type"]))

    # Behaviour-based items
    if row["scrolled_to_reviews"] == 1:
        transaction.append("Scrolled_Reviews")

    if row["added_to_cart"] == 1:
        transaction.append("Added_To_Cart")

    if row["converted"] == 1:
        transaction.append("Converted")

    if row["previous_app_user"] == 1:
        transaction.append("Previous_App_User")

    if row["previous_product_buyer"] == 1:
        transaction.append("Previous_Product_Buyer")

    transactions.append(transaction)


# ============================================================
# 3. DISPLAY SAMPLE TRANSACTIONS
# ============================================================

print("\nSample Transactions:")

for transaction in transactions[:5]:
    print(transaction)


# ============================================================
# 4. CREATE TID SETS
# ============================================================

item_tid = defaultdict(set)

for tid, transaction in enumerate(transactions):

    for item in transaction:
        item_tid[item].add(tid)


# ============================================================
# 5. ECLAT FUNCTION
# ============================================================

def eclat(prefix, items, min_support_count, results):

    while items:

        item, tidset = items.pop()

        support_count = len(tidset)

        if support_count >= min_support_count:

            new_itemset = prefix + [item]

            results.append(
                (
                    tuple(sorted(new_itemset)),
                    support_count
                )
            )

            suffix = []

            for other_item, other_tidset in items:

                intersection = tidset & other_tidset

                if len(intersection) >= min_support_count:

                    suffix.append(
                        (
                            other_item,
                            intersection
                        )
                    )

            eclat(
                new_itemset,
                suffix,
                min_support_count,
                results
            )


# ============================================================
# 6. SET MINIMUM SUPPORT
# ============================================================

min_support = 0.05

min_support_count = int(
    len(transactions) * min_support
)

print("\nMinimum Support:", min_support)

print(
    "Minimum Support Count:",
    min_support_count
)


# ============================================================
# 7. RUN ECLAT
# ============================================================

results = []

items = list(item_tid.items())

eclat(
    prefix=[],
    items=items,
    min_support_count=min_support_count,
    results=results
)


# ============================================================
# 8. CREATE RESULT DATAFRAME
# ============================================================

result_df = pd.DataFrame(
    results,
    columns=[
        "Itemset",
        "Support_Count"
    ]
)


# ============================================================
# 9. CALCULATE SUPPORT %
# ============================================================

result_df["Support"] = (
    result_df["Support_Count"]
    / len(transactions)
)

result_df["Support_Percentage"] = (
    result_df["Support"] * 100
)


# ============================================================
# 10. SORT BY SUPPORT
# ============================================================

result_df = result_df.sort_values(
    by="Support",
    ascending=False
).reset_index(drop=True)


# ============================================================
# 11. DISPLAY FREQUENT ITEMSETS
# ============================================================

print("\n============================================")
print("FREQUENT ITEMSETS USING ECLAT")
print("============================================")

print(
    result_df.head(20).to_string(index=False)
)


# ============================================================
# 12. BAR CHART - TOP 15 ITEMSETS
# ============================================================

top_15 = result_df.head(15).copy()

top_15["Itemset_Label"] = (
    top_15["Itemset"]
    .apply(lambda x: " + ".join(x))
)

plt.figure(figsize=(12, 7))

plt.barh(
    top_15["Itemset_Label"],
    top_15["Support_Percentage"]
)

plt.xlabel("Support (%)")

plt.ylabel("Itemset")

plt.title(
    "Top 15 Frequent Itemsets - ECLAT"
)

plt.gca().invert_yaxis()

plt.tight_layout()

plt.show()


# ============================================================
# 13. LINE CHART - TOP 15 ITEMSETS
# ============================================================

plt.figure(figsize=(12, 6))

plt.plot(
    range(1, len(top_15) + 1),
    top_15["Support_Percentage"],
    marker="o",
    linewidth=2
)

plt.xlabel("Itemset Rank")

plt.ylabel("Support (%)")

plt.title(
    "Support Trend of Top Frequent Itemsets"
)

plt.grid(True)

plt.tight_layout()

plt.show()


# ============================================================
# 14. SCATTER CHART
# ============================================================

itemset_sizes = result_df["Itemset"].apply(len)

plt.figure(figsize=(8, 6))

plt.scatter(
    itemset_sizes,
    result_df["Support_Percentage"],
    alpha=0.6
)

plt.xlabel("Itemset Size")

plt.ylabel("Support (%)")

plt.title(
    "Itemset Size vs Support"
)

plt.grid(True)

plt.tight_layout()

plt.show()


# ============================================================
# 15. ITEM FREQUENCY
# ============================================================

single_items = result_df[
    result_df["Itemset"].apply(len) == 1
].copy()

single_items["Item"] = (
    single_items["Itemset"]
    .apply(lambda x: x[0])
)

single_items = single_items.sort_values(
    by="Support",
    ascending=False
).head(15)


# ============================================================
# 16. ITEM FREQUENCY BAR CHART
# ============================================================

plt.figure(figsize=(10, 6))

plt.bar(
    single_items["Item"],
    single_items["Support_Percentage"]
)

plt.xlabel("Items")

plt.ylabel("Support (%)")

plt.title(
    "Most Frequent Individual Items"
)

plt.xticks(
    rotation=45,
    ha="right"
)

plt.tight_layout()

plt.show()


# ============================================================
# 17. HEATMAP - TOP ITEMS
# ============================================================

top_items = single_items["Item"].tolist()

co_matrix = pd.DataFrame(
    0,
    index=top_items,
    columns=top_items
)

for transaction in transactions:

    current_items = [
        item
        for item in transaction
        if item in top_items
    ]

    for item1, item2 in combinations(
        current_items,
        2
    ):

        co_matrix.loc[item1, item2] += 1
        co_matrix.loc[item2, item1] += 1


plt.figure(figsize=(10, 8))

sns.heatmap(
    co_matrix,
    annot=True,
    fmt="d",
    cmap="Blues"
)

plt.title(
    "Item Co-occurrence Heatmap"
)

plt.tight_layout()

plt.show()


# ============================================================
# 18. SAVE RESULTS
# ============================================================

result_df.to_csv(
    "eclat_frequent_itemsets.csv",
    index=False
)

print(
    "\nResults saved as: eclat_frequent_itemsets.csv"
)
```

```python
# ============================================================
# APRIORI ASSOCIATION RULE MINING
# Tesco Groceries Dataset
# Unsupervised Machine Learning
# ============================================================

import pandas as pd
import matplotlib.pyplot as plt

from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules

# ============================================================
# 1. LOAD DATASET
# ============================================================

df = pd.read_csv("tesco_groceries_dataset.csv")

print("Dataset Loaded Successfully!")
print("Shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nFirst 5 Rows:")
print(df.head())

# ============================================================
# 2. CHECK BREADCRUMBS
# ============================================================

print("\nSample Categories:")

print(
    df["breadcrumbs"]
    .dropna()
    .head()
)

# ============================================================
# 3. CREATE TRANSACTIONS
# ============================================================

# Each product's breadcrumbs are treated
# as one transaction containing category items.

transactions = []

for value in df["breadcrumbs"].dropna():

    items = [
        item.strip()
        for item in str(value).split("~")
        if item.strip()
    ]

    if len(items) >= 2:
        transactions.append(items)

print("\nTotal Transactions:", len(transactions))

# ============================================================
# 4. TRANSACTION ENCODING
# ============================================================

te = TransactionEncoder()

te_array = te.fit(
    transactions
).transform(
    transactions
)

basket = pd.DataFrame(
    te_array,
    columns=te.columns_
)

print("\nTransaction Matrix:")
print(basket.head())

# ============================================================
# 5. APRIORI - FREQUENT ITEMSETS
# ============================================================

frequent_itemsets = apriori(
    basket,
    min_support=0.05,
    use_colnames=True
)

# Add item count

frequent_itemsets["item_count"] = (
    frequent_itemsets["itemsets"]
    .apply(len)
)

# Sort by support

frequent_itemsets = frequent_itemsets.sort_values(
    by="support",
    ascending=False
)

print("\n======================================")
print("FREQUENT ITEMSETS")
print("======================================")

print(
    frequent_itemsets.head(20)
)

# ============================================================
# 6. ASSOCIATION RULES
# ============================================================

rules = association_rules(
    frequent_itemsets,
    metric="confidence",
    min_threshold=0.3
)

# ============================================================
# 7. FILTER RULES
# ============================================================

rules = rules[
    rules["lift"] > 1
]

# Sort by lift

rules = rules.sort_values(
    by="lift",
    ascending=False
)

print("\n======================================")
print("ASSOCIATION RULES")
print("======================================")

print(
    rules[
        [
            "antecedents",
            "consequents",
            "support",
            "confidence",
            "lift"
        ]
    ].head(20)
)

# ============================================================
# 8. CONVERT SETS TO STRINGS
# ============================================================

rules["antecedents"] = rules[
    "antecedents"
].apply(
    lambda x: ", ".join(list(x))
)

rules["consequents"] = rules[
    "consequents"
].apply(
    lambda x: ", ".join(list(x))
)

# ============================================================
# 9. TOP FREQUENT ITEMSETS
# ============================================================

top_itemsets = frequent_itemsets.head(10).copy()

top_itemsets["items"] = top_itemsets[
    "itemsets"
].apply(
    lambda x: ", ".join(list(x))
)

# ============================================================
# 10. VISUALIZATION 1
# TOP FREQUENT ITEMSETS
# ============================================================

plt.figure(figsize=(12, 6))

plt.bar(
    top_itemsets["items"],
    top_itemsets["support"]
)

plt.xlabel("Itemsets")
plt.ylabel("Support")

plt.title(
    "Top 10 Frequent Itemsets - Apriori"
)

plt.xticks(
    rotation=45,
    ha="right"
)

plt.tight_layout()

plt.show()

# ============================================================
# 11. VISUALIZATION 2
# SUPPORT VS CONFIDENCE
# ============================================================

plt.figure(figsize=(9, 6))

plt.scatter(
    rules["support"],
    rules["confidence"],
    s=50
)

plt.xlabel("Support")
plt.ylabel("Confidence")

plt.title(
    "Association Rules - Support vs Confidence"
)

plt.grid(True)

plt.show()

# ============================================================
# 12. VISUALIZATION 3
# CONFIDENCE VS LIFT
# ============================================================

plt.figure(figsize=(9, 6))

plt.scatter(
    rules["confidence"],
    rules["lift"],
    s=50
)

plt.xlabel("Confidence")
plt.ylabel("Lift")

plt.title(
    "Association Rules - Confidence vs Lift"
)

plt.grid(True)

plt.show()

# ============================================================
# 13. TOP 10 RULES BY LIFT
# ============================================================

top_rules = rules.head(10).copy()

top_rules["rule"] = (
    top_rules["antecedents"]
    + " → "
    + top_rules["consequents"]
)

plt.figure(figsize=(12, 6))

plt.bar(
    top_rules["rule"],
    top_rules["lift"]
)

plt.xlabel("Association Rules")
plt.ylabel("Lift")

plt.title(
    "Top 10 Association Rules by Lift"
)

plt.xticks(
    rotation=45,
    ha="right"
)

plt.tight_layout()

plt.show()

# ============================================================
# 14. DISPLAY BEST RULE
# ============================================================

if len(rules) > 0:

    best_rule = rules.iloc[0]

    print("\n======================================")
    print("BEST ASSOCIATION RULE")
    print("======================================")

    print(
        "Rule:",
        best_rule["antecedents"],
        "→",
        best_rule["consequents"]
    )

    print(
        "Support:",
        round(best_rule["support"], 4)
    )

    print(
        "Confidence:",
        round(best_rule["confidence"], 4)
    )

    print(
        "Lift:",
        round(best_rule["lift"], 4)
    )

# ============================================================
# 15. SAVE RESULTS
# ============================================================

frequent_itemsets.to_csv(
    "frequent_itemsets.csv",
    index=False
)

rules.to_csv(
    "association_rules.csv",
    index=False
)

print("\nResults saved successfully!")

print("1. frequent_itemsets.csv")
print("2. association_rules.csv")

# ============================================================
# 16. FINAL SUMMARY
# ============================================================

print("\n======================================")
print("APRIORI COMPLETED")
print("======================================")

print(
    "Transactions:",
    len(transactions)
)

print(
    "Frequent Itemsets:",
    len(frequent_itemsets)
)

print(
    "Association Rules:",
    len(rules)
)
```

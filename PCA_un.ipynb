```python
# ============================================================
# PCA - WINE QUALITY DATASET
# UNSUPERVISED LEARNING
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA


# ============================================================
# 1. LOAD DATASET
# ============================================================

df = pd.read_csv("winequalityN.csv")

print("Dataset Shape:", df.shape)

print("\nFirst 5 Rows:")
print(df.head())


# ============================================================
# 2. CHECK MISSING VALUES
# ============================================================

print("\nMissing Values:")
print(df.isnull().sum())


# ============================================================
# 3. HANDLE MISSING VALUES
# ============================================================

# Numeric columns -> median
numeric_columns = df.select_dtypes(
    include=np.number
).columns

df[numeric_columns] = df[numeric_columns].fillna(
    df[numeric_columns].median()
)


# Categorical column -> mode
df["type"] = df["type"].fillna(
    df["type"].mode()[0]
)


# ============================================================
# 4. ENCODE CATEGORICAL COLUMN
# ============================================================

df["type"] = df["type"].map({
    "white": 0,
    "red": 1
})


# ============================================================
# 5. SELECT FEATURES
# ============================================================

# Quality is NOT used because PCA is unsupervised.
# It is only kept separately for optional visualization.

X = df.drop(
    columns=["quality"]
)

y = df["quality"]


# ============================================================
# 6. STANDARDIZE FEATURES
# ============================================================

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)


# ============================================================
# 7. APPLY PCA
# ============================================================

pca = PCA()

X_pca = pca.fit_transform(
    X_scaled
)


# ============================================================
# 8. EXPLAINED VARIANCE
# ============================================================

explained_variance = (
    pca.explained_variance_ratio_ * 100
)

cumulative_variance = np.cumsum(
    explained_variance
)


print("\n========================================")
print("PCA EXPLAINED VARIANCE")
print("========================================")

for i, variance in enumerate(
    explained_variance
):

    print(
        f"PC{i + 1}: {variance:.2f}%"
    )

print(
    f"\nTotal Variance Explained: "
    f"{explained_variance.sum():.2f}%"
)


# ============================================================
# 9. PCA RESULT DATAFRAME
# ============================================================

pca_columns = [
    f"PC{i + 1}"
    for i in range(X_pca.shape[1])
]

pca_df = pd.DataFrame(
    X_pca,
    columns=pca_columns
)

print("\nPCA Data:")
print(pca_df.head())


# ============================================================
# 10. BAR CHART - EXPLAINED VARIANCE
# ============================================================

plt.figure(figsize=(10, 6))

plt.bar(
    range(1, len(explained_variance) + 1),
    explained_variance
)

plt.xlabel("Principal Components")

plt.ylabel("Explained Variance (%)")

plt.title(
    "PCA - Explained Variance by Principal Component"
)

plt.xticks(
    range(1, len(explained_variance) + 1)
)

plt.tight_layout()

plt.show()


# ============================================================
# 11. LINE CHART - EXPLAINED VARIANCE
# ============================================================

plt.figure(figsize=(10, 6))

plt.plot(
    range(1, len(explained_variance) + 1),
    explained_variance,
    marker="o",
    linewidth=2
)

plt.xlabel("Principal Components")

plt.ylabel("Explained Variance (%)")

plt.title(
    "PCA Scree Plot"
)

plt.xticks(
    range(1, len(explained_variance) + 1)
)

plt.grid(True)

plt.tight_layout()

plt.show()


# ============================================================
# 12. LINE CHART - CUMULATIVE VARIANCE
# ============================================================

plt.figure(figsize=(10, 6))

plt.plot(
    range(1, len(cumulative_variance) + 1),
    cumulative_variance,
    marker="o",
    linewidth=2
)

plt.axhline(
    y=90,
    linestyle="--",
    label="90% Variance"
)

plt.xlabel("Number of Principal Components")

plt.ylabel(
    "Cumulative Explained Variance (%)"
)

plt.title(
    "PCA - Cumulative Explained Variance"
)

plt.xticks(
    range(1, len(cumulative_variance) + 1)
)

plt.legend()

plt.grid(True)

plt.tight_layout()

plt.show()


# ============================================================
# 13. 2D PCA SCATTER PLOT
# ============================================================

plt.figure(figsize=(10, 7))

scatter = plt.scatter(
    pca_df["PC1"],
    pca_df["PC2"],
    c=y,
    alpha=0.6
)

plt.xlabel("Principal Component 1")

plt.ylabel("Principal Component 2")

plt.title(
    "PCA 2D Visualization - Wine Dataset"
)

plt.colorbar(
    scatter,
    label="Wine Quality"
)

plt.grid(True)

plt.tight_layout()

plt.show()


# ============================================================
# 14. 3D PCA SCATTER PLOT
# ============================================================

from mpl_toolkits.mplot3d import Axes3D

fig = plt.figure(
    figsize=(10, 7)
)

ax = fig.add_subplot(
    111,
    projection="3d"
)

scatter = ax.scatter(
    pca_df["PC1"],
    pca_df["PC2"],
    pca_df["PC3"],
    c=y,
    alpha=0.6
)

ax.set_xlabel("PC1")

ax.set_ylabel("PC2")

ax.set_zlabel("PC3")

ax.set_title(
    "PCA 3D Visualization - Wine Dataset"
)

fig.colorbar(
    scatter,
    ax=ax,
    label="Wine Quality"
)

plt.tight_layout()

plt.show()


# ============================================================
# 15. PCA COMPONENT LOADINGS
# ============================================================

loadings = pd.DataFrame(
    pca.components_.T,
    columns=pca_columns,
    index=X.columns
)

print("\n========================================")
print("PCA COMPONENT LOADINGS")
print("========================================")

print(
    loadings
)


# ============================================================
# 16. PC1 FEATURE CONTRIBUTION
# ============================================================

pc1 = loadings["PC1"].abs().sort_values(
    ascending=False
)

plt.figure(figsize=(10, 6))

plt.bar(
    pc1.index,
    pc1.values
)

plt.xlabel("Features")

plt.ylabel(
    "Absolute Loading"
)

plt.title(
    "Feature Contribution to PC1"
)

plt.xticks(
    rotation=45,
    ha="right"
)

plt.tight_layout()

plt.show()


# ============================================================
# 17. PC2 FEATURE CONTRIBUTION
# ============================================================

pc2 = loadings["PC2"].abs().sort_values(
    ascending=False
)

plt.figure(figsize=(10, 6))

plt.bar(
    pc2.index,
    pc2.values
)

plt.xlabel("Features")

plt.ylabel(
    "Absolute Loading"
)

plt.title(
    "Feature Contribution to PC2"
)

plt.xticks(
    rotation=45,
    ha="right"
)

plt.tight_layout()

plt.show()


# ============================================================
# 18. SAVE PCA DATA
# ============================================================

pca_df.to_csv(
    "wine_pca_results.csv",
    index=False
)

print(
    "\nPCA results saved as: wine_pca_results.csv"
)
```
